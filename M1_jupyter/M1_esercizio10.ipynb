# Procedura di consegna di un esercizio

**Destinatario:** un compagno di corso che deve consegnare un esercizio sul repository remoto senza chiedere spiegazioni a voce.

**Scopo:** fissare l'ordine dei passi dalla modifica del file fino al push, indicando per ciascuno perché esiste e che cosa non arriva al docente se quel passo viene saltato. I comandi da digitare stanno nei recinti bash; le celle di codice non eseguono Git, calcolano solo alcuni dati della procedura.


## 1. Modificare e salvare il file dell'esercizio

Questo passo esiste perché il repository può ricevere solo bytes già scritti su disco. La modifica lasciata aperta nell'editor, senza salvataggio, non è una modifica del file.

Se lo salti, i comandi successivi operano sulla versione precedente: il commit può riuscire e il push può riuscire, ma online resta il testo vecchio oppure un file vuoto. Prima di proseguire conviene anche controllare che il percorso sia quello richiesto dalla consegna.

```bash
ls M1_markdown/algoritmo.md
```


````python
numero_passi = 5
print(numero_passi)
````

## 2. Controllare lo stato del repository

Questo passo esiste perché la cartella di lavoro e l'area di stage non sono la stessa cosa. `git status` mostra che cosa è modificato, che cosa è già pronto per il commit e se il ramo locale è avanti rispetto al remoto.

Se lo salti, non hai un controllo sul contenuto della consegna: puoi dimenticare il file dell'esercizio oppure includere un file di prova che non andava inviato. L'errore si vede solo dopo, sul repository remoto.

```bash
git status
```


````python
minuti_per_passo = 2
numero_passi = 5
minuti_totali = minuti_per_passo * numero_passi
print(minuti_totali)
````

## 3. Portare il file nell'area di stage

Questo passo esiste perché un commit non legge tutte le modifiche della cartella: legge solo ciò che è stato messo in stage. `git add` con il percorso del file è la scelta più sicura, perché non trascina altri file aperti per sbaglio.

Se lo salti, il commit successivo non ha quel file nella fotografia. Il messaggio può essere corretto e il push può andare a buon fine, ma l'esercizio non è nel commit nuovo.

```bash
git add M1_markdown/algoritmo.md
```


````python
file_nel_commit = 1
print(file_nel_commit)
````

## 4. Creare il commit con un messaggio chiaro

Questo passo esiste perché il commit è la fotografia con un messaggio, ed è l'unità che il remoto potrà ricevere. Il messaggio deve dire quale esercizio viene consegnato, così nella storia del repository si distingue questa consegna dalle altre.

Se lo salti, non nasce un nuovo commit. Il push seguente non ha nulla di nuovo da inviare, oppure risponde che tutto è già aggiornato, e il docente continua a vedere la consegna precedente.

```bash
git commit -m "Consegna esercizio: algoritmo di ricerca sequenziale"
```


````python
comandi_git = 4
minuti_verifica = 3
minuti_comandi = 8
minuti_con_verifica = minuti_comandi + minuti_verifica
print(minuti_con_verifica)
````

## 5. Inviare i commit al repository remoto

Questo passo esiste perché la correzione guarda il repository remoto, non il computer su cui hai lavorato. `git push` copia sul remoto i commit che esistono solo in locale. Se il ramo del corso non si chiama `main`, va usato il nome reale del ramo, quello mostrato da `git status`.

Se lo salti, il lavoro resta sul computer locale. Il commit c'è, il messaggio c'è, ma sulla pagina GitHub l'ultimo aggiornamento non cambia e l'esercizio risulta non consegnato.

```bash
git push origin main
```

## Comandi e effetto

| Comando | Effetto |
| --- | --- |
| `ls M1_markdown/algoritmo.md` | conferma che il file salvato è nel percorso atteso |
| `git status` | elenca modifiche, file in stage e rapporto col remoto |
| `git add M1_markdown/algoritmo.md` | mette quel file nell'area di stage |
| `git commit -m "..."` | crea un commit nuovo con il messaggio indicato |
| `git push origin main` | invia i commit locali al ramo remoto |


## Regola di verifica prima della consegna

> Prima di considerare consegnato l'esercizio, apri il repository su GitHub, controlla l'ultimo commit e verifica che il file sia proprio quello modificato. Se il file non compare online, o compare senza le modifiche, la procedura non è finita: manca il salvataggio, lo stage, il commit o il push.
